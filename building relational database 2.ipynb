### creating database

```python
cur.execute('''DROP TABLE schools''')
cur.execute('''DROP TABLE enrollment''')
```

```python
cur.execute('''
    CREATE TABLE schools (
        school_id INTEGER PRIMARY KEY AUTOINCREMENT,
        StateABR TEXT,
        SchoolName TEXT,
        Street1 TEXT,
        Street2 TEXT,
        City TEXT,
        State TEXT,
        Zip TEXT,
        Zip4 TEXT,
        SchoolLevel TEXT,
        Locale TEXT,
        County TEXT
    )
''')

cur.execute('''
    CREATE TABLE enrollment (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        SchoolName TEXT,
        Status TEXT,
        SchoolType TEXT,
        Status_Text TEXT,
        TotalEnrollment INTEGER
    )
''')

conn.commit()
```

```python
filename = "Public_School_Characteristics_2022-23.csv"

with open(filename) as csv_file:
    csv_reader = csv.reader(csv_file, delimiter=',')
    for row in csv_reader:
        print(row)
        StateABR = row[5]
        SchoolName = row[9]
        Street1 = row[10]
        Street2 = row[11]
        City = row[12]
        State = row[13]
        Zip = row[14] 
        Zip4 = row[15]
        SchoolLevel = row[21]
        Status = row[22]
        SchoolType = row[23]
        Status_Text = row[24]
        Locale = row[25]
        County = row[26]
        TotalEnrollment = row[50]

        cur.execute('''
            INSERT INTO schools (StateABR, SchoolName, Street1, Street2, 
            City, State, Zip, Zip4, SchoolLevel, Locale, County)
            VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?) ''',
                    (StateABR, SchoolName, Street1, Street2, City, State, Zip, Zip4, SchoolLevel, Locale, County))

        
        cur.execute('''
            INSERT INTO enrollment (SchoolName, Status, SchoolType, 
            Status_Text, TotalEnrollment)
            VALUES (?, ?, ?, ?, ?) ''',
                    (SchoolName, Status, SchoolType, Status_Text, TotalEnrollment))


    conn.commit()
```

```python
cur.execute('''
SELECT *
FROM schools
LIMIT 10;
''')

cur.fetchall()
```

```python
cur.execute('''
SELECT *
FROM enrollment
WHERE TotalEnrollment > 100
LIMIT 10;
''')

cur.fetchall()
```

### Joining Tables

```python
cur.execute('''
SELECT 
    schools.SchoolName, 
    schools.City, 
    schools.State, 
    enrollment.Status, 
    enrollment.SchoolType, 
    enrollment.TotalEnrollment
FROM schools
INNER JOIN enrollment 
ON schools.SchoolName = enrollment.SchoolName;
''')

cur.fetchall()
```

### Querying Data

```python
cur.execute('''
SELECT SchoolName, City, State 
FROM schools
WHERE State = 'CA';
''')

cur.fetchall()
```

```python
cur.execute('''
SELECT SchoolName, TotalEnrollment
FROM enrollment
WHERE TotalEnrollment 
BETWEEN 1 AND 100;
''')

cur.fetchall()
```

### Removing Null Values

```python
cur.execute('''
DELETE
FROM enrollment 
WHERE TotalEnrollment = 'NULL' OR Status = 'NULL' OR SchoolType = 'NULL'
''')

cur.execute('''
DELETE
FROM schools 
WHERE City = 'NULL' OR State = 'NULL' OR SchoolName = 'NULL' OR StateABR = 'NULL' OR Street1 = 'NULL' OR County = 'NULL'
''')
```
