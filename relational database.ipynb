```python
import sqlite3
import csv
import numpy as np
import pandas as pd
```

```python
#creating connection & cursor
conn = sqlite3.connect('detect_types.sqlite') 
cur = conn.cursor()
```

```python
cur.execute('''
DROP TABLE schools
''')
```

```python
cur.execute('''
CREATE TABLE schools(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    city TEXT,
    enrollment INTEGER
    )
    ''')
```

```python
data = [('Vasquez High','Acton', 359),
        ('Manor Elementary','Fairfax', 236),
        ('Brookside Elementary','San Anselmo', 281),
        ('Cuyama Valley High','New Cuyama', 59),
        ('Mt. Lassen Charter','Herlong', 147),
        ('Grizzly Hill','Nevada City', 94),
        ('Antelope Creek Elementary','Rocklin', 472),
        ('Citrus Hill High','Perris', 1731),
        ('Lasselle Elementary','Moreno Valley', 717),
        ('Don Pedro High','La Grange', 50),
        ('Hollyvale Elementary','Victorville', 484),
       ('Mountain View High','Lucerne Valley',9)]

cur.executemany('''
INSERT INTO schools (name, city, enrollment) 
VALUES (?, ?, ?)
''',data)
conn.commit()
```

```python
#inserting into the table
cur.execute('''
INSERT INTO schools (name, city, enrollment)
VALUES ('Lucerne Valley Middle', 'Lucerne Valley', 137)
''')
conn.commit()
```
